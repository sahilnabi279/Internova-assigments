## Task 1: NumPy Fundamentals

NumPy (Numerical Python) is a Python library used for numerical computing.
It provides powerful data structures called arrays and functions for performing
mathematical and statistical operations efficiently.

NumPy is widely used in Artificial Intelligence, Machine Learning, Data Science,
and scientific computing.

```python
import numpy as np

print("NumPy imported successfully!")
```

### Creating a 1D NumPy Array

```python
array_1d = np.array([10, 20, 30, 40, 50])

print("1D Array:")
print(array_1d)
```

### Creating a 2D NumPy Array

```python
array_2d = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("2D Array:")
print(array_2d)
```

### Array Indexing

```python
print("First element:", array_1d[0])
print("Third element:", array_1d[2])

print("First row of 2D array:", array_2d[0])
print("Element at row 2, column 3:", array_2d[1, 2])
```

### Array Slicing

```python
print("First three elements:", array_1d[0:3])
print("Last two elements:", array_1d[3:5])
print("Every second element:", array_1d[::2])
```

### Shape and Dimensions

```python
print("Shape of 1D array:", array_1d.shape)
print("Dimensions of 1D array:", array_1d.ndim)

print("Shape of 2D array:", array_2d.shape)
print("Dimensions of 2D array:", array_2d.ndim)
```

### Mathematical Operations 

```python
numbers = np.array([10, 20, 30, 40, 50])

print("Array:", numbers)
print("Mean:", np.mean(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))
print("Sum:", np.sum(numbers))
```

### Basic Array Mathematical Operations

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

## Task 1 Conclusion

In this task, NumPy was used to create and manipulate one-dimensional and
two-dimensional arrays. Array indexing and slicing were performed to access
specific elements and portions of arrays. The shape and dimensions of arrays
were also examined. Mathematical functions such as mean, maximum, minimum,
and sum were used to perform basic statistical calculations. These operations
are important in AI and data science because numerical datasets are frequently
represented and processed using NumPy arrays.

# Task 3: Data Cleaning

Data cleaning is the process of identifying and correcting inaccurate,
incomplete, duplicate, or incorrectly formatted data.

In this task, a dataset containing missing values and duplicate records
will be analyzed and cleaned using Pandas.

```python
import pandas as pd

messy_df = pd.read_csv("messy_student_data.csv")

print("Messy dataset loaded successfully!")
```

```python
messy_df
```

```python
messy_df.info()
```

```python
messy_df.isnull()
```

```python
total_missing = messy_df.isnull().sum().sum()

print("Total missing values:", total_missing)
```

```python
print("Number of duplicate rows:", messy_df.duplicated().sum())
```

```python
messy_df[messy_df.duplicated()]
```

```python
print(messy_df.dtypes)
```

```python
messy_df["Study_Hours"] = messy_df["Study_Hours"].astype("string")
```

```python
print(messy_df.dtypes)
```

```python
messy_df["Study_Hours"] = pd.to_numeric(
    messy_df["Study_Hours"],
    errors="coerce"
)
```

```python
print("Study Hours Median:", messy_df["Study_Hours"].median())
print("Attendance Median:", messy_df["Attendance"].median())
print("Science Median:", messy_df["Science"].median())
```

```python
messy_df["Study_Hours"] = messy_df["Study_Hours"].fillna(
    messy_df["Study_Hours"].median()
)

messy_df["Attendance"] = messy_df["Attendance"].fillna(
    messy_df["Attendance"].median()
)

messy_df["Science"] = messy_df["Science"].fillna(
    messy_df["Science"].median()
)
```

```python
print(messy_df.isnull().sum())
```

```python
print("Total missing values:", messy_df.isnull().sum().sum())
```

```python
messy_df = messy_df.drop_duplicates()
```

```python
print("Number of duplicate rows:", messy_df.duplicated().sum())
```

```python
messy_df = messy_df.reset_index(drop=True)
```

```python
messy_df
```

```python
print("Dataset shape:", messy_df.shape)
print("\nMissing values:")
print(messy_df.isnull().sum())

print("\nDuplicate rows:", messy_df.duplicated().sum())

print("\nData types:")
print(messy_df.dtypes)
```

```python
messy_df[messy_df["Math"].isnull()]
```

```python
math_median = messy_df["Math"].median()

print("Math median:", math_median)

messy_df["Math"] = messy_df["Math"].fillna(math_median)
```

```python
print("Missing values after cleaning:")
print(messy_df.isnull().sum())

print("\nTotal missing values:",
      messy_df.isnull().sum().sum())
```

```python
print("Duplicate rows:", messy_df.duplicated().sum())
```

```python
messy_df["Study_Hours"] = messy_df["Study_Hours"].astype(int)
```

```python
print(messy_df.dtypes)
```

```python
print("===== FINAL CLEANING REPORT =====")

print("Dataset shape:", messy_df.shape)

print("\nMissing values:")
print(messy_df.isnull().sum())

print("\nTotal missing values:",
      messy_df.isnull().sum().sum())

print("\nDuplicate rows:",
      messy_df.duplicated().sum())

print("\nData types:")
print(messy_df.dtypes)
```

```python
messy_df
```

```python
messy_df.to_csv("cleaned_student_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```

# Task 4: Data Visualization

Data visualization is the graphical representation of data. It helps us
identify patterns, trends, distributions, and relationships within a dataset.

In this task, Matplotlib is used to create four different visualizations
from the cleaned student performance dataset.

```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd

df = pd.read_csv("cleaned_student_data.csv")

print("Cleaned dataset loaded successfully!")
```

```python
df.head()
```

# Bar Chart

```python
subjects = ["Math", "Science", "English"]

average_marks = [
    df["Math"].mean(),
    df["Science"].mean(),
    df["English"].mean()
]

plt.figure(figsize=(8, 5))

plt.bar(subjects, average_marks)

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.ylim(0, 100)

plt.show()
```

### Bar Chart Explanation

The bar chart compares the average marks obtained by students in Mathematics,
Science, and English. It allows us to easily compare performance across the
three subjects and identify which subject has the highest and lowest average.

# Line Chart

```python
plt.figure(figsize=(10, 5))

plt.plot(
    df["Name"],
    df["Math"],
    marker="o"
)

plt.title("Mathematics Marks by Student")
plt.xlabel("Student")
plt.ylabel("Math Marks")

plt.xticks(rotation=45)

plt.ylim(0, 100)

plt.tight_layout()

plt.show()
```

### Line Chart Explanation

The line chart shows the Mathematics marks of individual students. It helps
visualize how performance changes from one student to another and makes
higher and lower marks easy to identify.

# Histogram

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Math"],
    bins=5,
    edgecolor="black"
)

plt.title("Distribution of Mathematics Marks")
plt.xlabel("Math Marks")
plt.ylabel("Number of Students")

plt.show()
```

### Histogram Explanation

The histogram shows the distribution of Mathematics marks among students.
It helps identify the range in which most students scored and provides an
overview of the spread of student performance.

# Scatter Plot

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study_Hours"],
    df["Math"]
)

plt.title("Study Hours vs Mathematics Marks")
plt.xlabel("Study Hours")
plt.ylabel("Math Marks")

plt.grid(True)

plt.show()
```

### Scatter Plot Explanation

The scatter plot shows the relationship between students' study hours and
their Mathematics marks. The graph can be used to observe whether students
who study for more hours generally tend to achieve higher Mathematics marks.

## Task 4 Conclusion

Four different types of visualizations were created using Matplotlib to
analyze the student performance dataset. The bar chart compared average
performance across subjects, while the line chart displayed Mathematics
marks for individual students. The histogram showed the distribution of
Mathematics marks, and the scatter plot examined the relationship between
study hours and Mathematics performance.

These visualizations make numerical data easier to understand and help
identify patterns, differences, distributions, and possible relationships
within the dataset. Data visualization is an important part of data analysis
because graphical representations can communicate insights more clearly than
raw numerical values.

# Task 5: Mini Project
## Student Performance Data Analysis

This mini project analyzes a student performance dataset using Python,
Pandas, NumPy, and Matplotlib. The analysis focuses on data handling,
cleaning, descriptive statistics, student pass/fail performance, and
visualization of important patterns.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cleaned_student_data.csv")

print("Dataset loaded successfully!")
```

```python
df
```

```python
print("Number of students:", len(df))
print("Number of columns:", len(df.columns))
```

```python
df.info()
```

```python
print("Dataset shape:", df.shape)
```

```python
important_columns = [
    "Name",
    "Study_Hours",
    "Attendance",
    "Math",
    "Science",
    "English"
]

print("Important columns:")
print(important_columns)
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("\nDuplicate records:", df.duplicated().sum())
```

```python
average_math = df["Math"].mean()
average_science = df["Science"].mean()
average_english = df["English"].mean()

print("Average Math Marks:", round(average_math, 2))
print("Average Science Marks:", round(average_science, 2))
print("Average English Marks:", round(average_english, 2))
```

```python
df["Average_Marks"] = df[["Math", "Science", "English"]].mean(axis=1)

overall_average = df["Average_Marks"].mean()

print("Overall Average Marks:", round(overall_average, 2))
```

```python
highest_math = df["Math"].max()

print("Highest Math Marks:", highest_math)
```

```python
highest_average = df["Average_Marks"].max()

print("Highest Overall Average:", round(highest_average, 2))
```

```python
top_student = df.loc[
    df["Average_Marks"].idxmax(),
    "Name"
]

print("Top Performing Student:", top_student)
```

```python
lowest_math = df["Math"].min()

print("Lowest Math Marks:", lowest_math)
```

```python
lowest_average = df["Average_Marks"].min()

print("Lowest Overall Average:", round(lowest_average, 2))
```

```python
lowest_student = df.loc[
    df["Average_Marks"].idxmin(),
    "Name"
]

print("Lowest Overall Performance:", lowest_student)
```

```python
PASS_MARK = 40

df["Result"] = np.where(
    df["Average_Marks"] >= PASS_MARK,
    "Pass",
    "Fail"
)
```

```python
df[["Name", "Average_Marks", "Result"]]
```

```python
passed_students = (df["Result"] == "Pass").sum()

print("Number of students who passed:", passed_students)
```

```python
failed_students = (df["Result"] == "Fail").sum()

print("Number of students who failed:", failed_students)
```

```python
print("STUDENT PERFORMANCE SUMMARY")

print("Total Students:", len(df))

print("Average Math Marks:", round(df["Math"].mean(), 2))
print("Average Science Marks:", round(df["Science"].mean(), 2))
print("Average English Marks:", round(df["English"].mean(), 2))

print("Overall Average Marks:", round(df["Average_Marks"].mean(), 2))

print("Highest Overall Average:",
      round(df["Average_Marks"].max(), 2))

print("Lowest Overall Average:",
      round(df["Average_Marks"].min(), 2))

print("Top Performing Student:", top_student)
print("Lowest Performing Student:", lowest_student)

print("Students Passed:", passed_students)
print("Students Failed:", failed_students)
```

```python
subjects = ["Math", "Science", "English"]

averages = [
    df["Math"].mean(),
    df["Science"].mean(),
    df["English"].mean()
]

plt.figure(figsize=(8, 5))

plt.bar(subjects, averages)

plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")

plt.ylim(0, 100)

plt.show()
```

### Chart 1: Average Marks by Subject

This bar chart compares the average marks achieved by students in
Mathematics, Science, and English. It helps identify differences in
overall performance across the three subjects.

```python
plt.figure(figsize=(10, 5))

plt.plot(
    df["Name"],
    df["Average_Marks"],
    marker="o"
)

plt.title("Average Marks of Students")
plt.xlabel("Student")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)

plt.ylim(0, 100)

plt.tight_layout()

plt.show()
```

### Chart 2: Average Marks of Students

The line chart shows the overall average marks of each student. It makes it
easy to compare individual performance and identify students with relatively
higher or lower overall averages.

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study_Hours"],
    df["Average_Marks"]
)

plt.title("Study Hours vs Average Marks")
plt.xlabel("Study Hours")
plt.ylabel("Average Marks")

plt.grid(True)

plt.show()
```

### Chart 3: Study Hours vs Average Marks

The scatter plot shows the relationship between students' study hours and
their overall average marks. It helps us observe whether students who spend
more time studying generally achieve higher marks.

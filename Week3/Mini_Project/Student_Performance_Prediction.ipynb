# Student Performance Prediction

## Week 3 - Machine Learning Mini Project

**Project:** Student Performance Prediction  
**Model:** Logistic Regression  
**Dataset:** Student Performance Dataset  
**Objective:** Predict whether a student is likely to PASS or FAIL based on academic and attendance-related features.

## 1. Project Objective

The objective of this mini-project is to build a simple machine learning classification system that predicts whether a student is likely to PASS or FAIL.

The model uses the following student-related features:

- Study Hours
- Attendance
- Previous Marks
- Assignment Score
- Project Score

A Logistic Regression classification model is trained using the Student Performance dataset. After training, the model is evaluated using accuracy and then tested with new student inputs.

## 2. Import Required Libraries

Pandas is used to load and manipulate the dataset. Scikit-learn is used for preprocessing, splitting the data, training the Logistic Regression model, and evaluating its performance.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

print("All required libraries imported successfully!")
```

 ## 3. Load the Dataset

The Student Performance dataset created during the previous tasks is loaded from `Dataset.csv`.

Using the same dataset maintains consistency between dataset preparation, preprocessing, model training, and this mini-project.

```python
df = pd.read_csv("../Dataset.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

## 4. Explore the Dataset

The first five rows of the dataset are displayed to verify that the dataset has been loaded correctly.

```python
df.head()
```

## 5. Data Cleaning

Before training the model, the dataset is checked for missing values. Missing values can cause problems during model training, so they must be identified and handled.

```python
print("Missing values:")
print(df.isnull().sum())
```

## 6. Select Features and Target

The input features are separated from the target variable.

The five features used by the model are:

- Study Hours
- Attendance
- Previous Marks
- Assignment Score
- Project Score

The target variable is `Result`, which contains PASS or FAIL.

```python
X = df.drop("Result", axis=1)
y = df["Result"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

## 7. Encode the Target Variable

The target variable contains categorical values, PASS and FAIL. These values are converted into numerical labels using LabelEncoder so that they can be processed by the machine learning model.

```python
encoder = LabelEncoder()

y = encoder.fit_transform(y)

print("Class mapping:")

for class_name, encoded_value in zip(
    encoder.classes_,
    encoder.transform(encoder.classes_)
):
    print(class_name, "=", encoded_value)
```

## 8. Train-Test Split

The dataset is divided into training and testing sets using an 80:20 ratio.

The training data is used to train the model, while the testing data is used to evaluate its performance on unseen student records.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

## 9. Data Preprocessing

StandardScaler is used to standardize the numerical features.

The scaler is fitted only on the training data and is then applied to the testing data. This prevents information from the testing dataset from being used during training.

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Preprocessing completed successfully!")
```

## 10. Train the Logistic Regression Model

Logistic Regression is used as the classification algorithm.

The model learns the relationship between student performance features and the target variable, which represents whether the student PASSes or FAILs.

```python
model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)

print("Model trained successfully!")
```

## 11. Make Predictions

The trained model is used to predict the results of students in the testing dataset.

Because the testing data was not used during training, it provides an estimate of how the model performs on unseen student records.

```python
y_pred = model.predict(X_test_scaled)

print("Predictions:")
print(y_pred)
```

## 12. Model Accuracy

Accuracy is used to measure the percentage of testing records that were classified correctly.

A higher accuracy means that the model correctly predicted a greater proportion of the unseen testing records.

```python
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
print("Model Accuracy (%):", accuracy * 100)
```

## 13. Classification Report

A classification report is generated to provide additional information about the performance of the model for the PASS and FAIL classes.

```python
print(classification_report(
    y_test,
    y_pred,
    target_names=encoder.classes_
))
```

## 14. Test the Model with New Student Inputs

To demonstrate how the trained model can be used in a real-world situation, three new student records are provided.

The model will use their study hours, attendance, previous marks, assignment score, and project score to predict whether each student is likely to PASS or FAIL.

```python
new_students = pd.DataFrame({
    "Study_Hours": [6, 3, 8],
    "Attendance": [85, 65, 95],
    "Previous_Marks": [72, 50, 88],
    "Assignment_Score": [80, 55, 92],
    "Project_Score": [78, 52, 90]
})

new_students
```

## 15. Preprocess New Student Data

The new student inputs must undergo the same preprocessing used for the training data. The previously fitted StandardScaler is therefore used to transform the new records.

```python
new_students_scaled = scaler.transform(new_students)

print("New student data preprocessed successfully!")
```

## 16. Predict New Student Results

The trained Logistic Regression model is used to predict the results of the three new students.

```python
new_predictions = model.predict(new_students_scaled)

predicted_results = encoder.inverse_transform(new_predictions)

new_students["Prediction"] = predicted_results

new_students
```

## 17. Prediction Probability

Logistic Regression can also provide the probability associated with each class. This gives an indication of how confident the model is about its prediction.

```python
probabilities = model.predict_proba(new_students_scaled)

probability_results = pd.DataFrame({
    "Prediction": predicted_results,
    "FAIL Probability (%)": probabilities[:, 0] * 100,
    "PASS Probability (%)": probabilities[:, 1] * 100
})

probability_results
```

## 18. Save the Trained Model

The trained Logistic Regression model and the StandardScaler are saved so that they can be reused later without retraining the model from the beginning.

```python
import joblib

joblib.dump(model, "student_performance_model.pkl")
joblib.dump(scaler, "student_performance_scaler.pkl")
joblib.dump(encoder, "student_performance_encoder.pkl")

print("Model and preprocessing objects saved successfully!")
```
